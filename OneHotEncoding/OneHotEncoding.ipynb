```python
#let's find the 10 most frequent categories in column X2 
df.X2.value_counts().sort_values(ascending=False).head(20)
```

```python
#you can see above that each category is repeated for the following above number of times
# now we will take top most frequent variables and make the rest as zero to avoid from high dimentionality
```

```python
# now let's make a list with the most frequent categories of the variable
top_10= [x for x in df.X2.value_counts().sort_values(ascending=False).head(10).index]
top_10
```

```python
# Now I will make the 10 binary variables
# it will find where label will be equal to column vlaue it will replace it with 1 else 0

for label in top_10:
    df[label]= np.where(df['X2']==label, 1,0)

df[['X2']+ top_10].head(20)
```

```python
# Now get whole set of dummy variables for all the categorical columns
def one_hot_encoding(df,col_name,top_X_labels):
    #Func to create the dummy variables for most frequest labels
    #we can vary the most frequent number of labels that we chosse to encode
    for label in top_X_labels:
        df[col_name+'_'+label]= np.where(df[col_name]==label, 1,0)

#read the data again
df= pd.read_csv('train.csv',usecols=['X1','X2','X3','X4','X5','X6','X8'])
#encode X2 into 10 most frequent categories
one_hot_encoding(df,'X2', top_10)

df.head()
```
