## Ordinal Number Encoding 
#### you assign your categories in such a manner where you can rank them. 
#### e.g. Grading System
#### where you rank on the basis of different grades (A,B,C,D,F).
#### Another example can be 'The Experience of Batsman'
#### Like how many number of Experience a batsman can have. As experience will increase he will probably played well.

```python
import datetime
```

```python
today_date= datetime.datetime.today()
today_date
```

```python
### List Comprehension
### it will give us the difference of todays date with the number of days we pass
days=[today_date-datetime.timedelta(x) for x in range(0,15)]
days
```

```python
## if I have to convert it into the dataframes
import pandas as pd
df= pd.DataFrame(days)
df.columns=['Days']
```

```python
df.head()
```

```python
df['WeekDay']=df['Days'].dt.day_name()
df.head()
```

```python
## Now I will make dictionary to encode these weekday column
dict= {
    'Monday':1,
    'Tuesday':2,
    'Wednesday':3,
    'Thursday':4,
    'Friday':5,
    'Saturday':6,
    'Sunday':7
}
dict
```

```python
### Ordinal Encoding
df['WeekDay_Ordinal']=df['WeekDay'].map(dict)
df['WeekDay_Ordinal']
```
